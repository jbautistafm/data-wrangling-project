## Scrapping de Maratones (Berlín y Londres)

```python
import os, sys
import pandas as pd
sys.path.append("../src")
from scraping import get_berlin, get_londres, get_clima_elite

elite = pd.concat([get_berlin(), get_londres()], ignore_index=True)
ruta = "../data/raw/elite_weather.csv"
if os.path.exists(ruta):
    clima_elite = pd.read_csv(ruta, parse_dates=["fecha"])
else:
    clima_elite = get_clima_elite(elite)
    clima_elite.to_csv(ruta, index=False)

elite_clima = elite.merge(clima_elite, on=["maraton", "fecha"], how="left")
# estos datos son publicos, los guardo en clean para usarlos en el eda
elite_clima.to_csv("../data/clean/elite.csv", index=False)
print(elite_clima.shape)
```
